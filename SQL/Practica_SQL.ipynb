## Groupby

```python
%%sql
SELECT status, COUNT(*) AS n
FROM orders
GROUP BY status
ORDER BY n DESC;
```

Aca vemos con el groupby para ver que hay de cada cosa

```python
%%sql
SELECT * FROM orders
LIMIT 5;
```

```python
%%sql
SELECT restaurant_id, order_ts,  COUNT(*) AS n_orders
FROM orders
GROUP BY restaurant_id
HAVING COUNT(*) >= 5
ORDER BY n_orders DESC;
```

## Subqueries

```python
%%sql
SELECT customer_id
FROM orders
GROUP BY customer_id
HAVING COUNT(*) >= 3;
```

```python
%%sql
SELECT *
FROM customers
WHERE customer_id IN (
    SELECT customer_id
    FROM orders
    GROUP BY customer_id
    HAVING COUNT(*) >= 3
);
```

Aca usamos un subquery para filtrar entonces para hacer los subqueries primero es definir el subquery. Entonces adentro esta pues lo que quremos relacionar con lo de arriba con lo del subquery pro ejemplo aca sacamos las persona que han hehco por lo menos 3 rdenes y las relacionamos con su infromacion con el IN

## JOINS

### INNER JOIN -> JOIN
Devuelve SOLO filas que tienen match en ambas tablas. -> Ambos el inner join y el join son lo mismo

### LEFT JOIN
todo lo de la tabla izquierda
Y lo que haga match en la derecha
Si no hay match → NULL

se puede usar para identificar cosas faltantes

### RIGHT JOIN

Es exactamente lo mismo que LEFT pero invertido.

### FULL OUTER JOIN

Todo lo de la izquierda
Todo lo de la derecha
Match cuando exista
NULL cuando no exista

### SELF JOIN

Una tabla unida consigo misma.

```python
%%sql
SELECT * FROM orders
inner join customers
on orders.customer_id = customers.customer_id
LIMIT 5;
```

INNER JOIN

```python
%%sql
SELECT o.order_id, o.status, c.courier_name
FROM orders o
LEFT JOIN couriers c
ON o.courier_id = c.courier_id
LIMIT 5;
```
